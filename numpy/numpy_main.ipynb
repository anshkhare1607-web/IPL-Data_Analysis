```python
# Task 1: Total Runs per Match

#unique match IDs from the dataset
unique_match_ids = np.unique(match_ids)

runs_per_match = []

# 2. Get the list of unique match IDs
unique_match_ids = np.unique(match_ids)

# 3. Loop through every match one by one
for m in unique_match_ids:
    # Create a mask: True for the rows belonging to match 'm', False otherwise
    mask = (match_ids == m)
    
    # Filter: Use the mask to pick only the runs for this specific match
    match_runs_only = batsman_runs[mask]
    
    # Aggregate: Sum all those filtered runs
    total_runs = np.sum(match_runs_only)
    
    # Store the match ID and the total as a tuple in our list
    runs_per_match.append((m, total_runs))


# 3. Printing the output in the specified format [(match_id, total_runs), ...]
print("Required Output Format:")
print(runs_per_match[:5], "...") # Showing first 5 matc hes

# 4. Printing a neat summary table for the first 10 matches
print(f"\n{'Match ID':<10} | {'Total Runs':<10}")
print("-" * 25)
for match_id, total_runs in runs_per_match[:10]:
    print(f"{int(match_id):<10} | {int(total_runs):<10}")
```

```python
# Task 5: Runs per Over (1 to 20)

# 1. Calculate Total Runs for each over (0-19)
total_runs_per_over = np.bincount(over, weights=batsman_runs)

# 2. Count how many times each over was bowled across all matches
match_over_combined = match_ids * 100 + over
unique_match_overs = np.unique(match_over_combined)

# Extract just the 'over' part from the unique combinations (0-19)
overs_bowled_list = unique_match_overs % 100
matches_per_over = np.bincount(overs_bowled_list)

# 3. Calculate Average: Total Runs / Number of Matches that bowled that over
avg_runs_per_over = total_runs_per_over / matches_per_over

# 4. Final Output Formatting
print("Average Runs Per Over (1 to 20):")
print(avg_runs_per_over)

# Neat Summary
print(f"\n{'Over Number':<12} | {'Average Runs':<15}")
print("-" * 30)
for i, avg in enumerate(avg_runs_per_over):
    # i + 1 because index 0 is actually the 1st over
    print(f"Over {i+1:<7} | {avg:<15.2f}")
```

```python
# Task 6: Boundary Analysis

# 1. Count total number of Fours (where batsman_runs is exactly 4)
fours_mask = (batsman_runs == 4)
total_fours = np.sum(fours_mask)

# 2. Count total number of Sixes (where batsman_runs is exactly 6)
sixes_mask = (batsman_runs == 6)
total_sixes = np.sum(sixes_mask)

# Bonus

# A. Create a mask for all boundaries 
boundary_mask = (batsman_runs == 4) | (batsman_runs == 6)

# B. Get unique teams and map every ball to a team ID
unique_teams, team_indices = np.unique(batting_team, return_inverse=True)

# C. Use bincount to sum up boundaries for each team
team_boundary_counts = np.bincount(team_indices, weights=boundary_mask)

# D. Find the team with the highest count 
most_boundaries_idx = np.argmax(team_boundary_counts)
team_with_most_boundaries = unique_teams[most_boundaries_idx]
max_boundaries_val = team_boundary_counts[most_boundaries_idx]

# --- Output ---
print(f"Total Fours: {total_fours}")
print(f"Total Sixes: {total_sixes}")
print(f"Total Boundaries: {total_fours + total_sixes}")

print(f"\n--- Bonus ---")
print(f"Team with most boundaries: {team_with_most_boundaries}")
print(f"Boundary Count: {int(max_boundaries_val)}")
```

```python
# Task 7: Death Overs Analysis 

# 1. Create a mask for death overs using combined boolean conditions
death_over_mask = (over >= 15) & (over <= 19) 

# 2. Calculate total runs scored in death overs
death_runs_total = np.sum(batsman_runs[death_over_mask])

# 3. Identify the team with the highest runs in death overs
death_teams = batting_team[death_over_mask]
death_runs_array = batsman_runs[death_over_mask]

# Get unique teams from the death overs data
u_death_teams, inv_death_indices = np.unique(death_teams, return_inverse=True)

# Sum runs for each team specifically in these overs
death_team_totals = np.bincount(inv_death_indices, weights=death_runs_array)

# Find the index of the highest scoring team
top_death_team_idx = np.argmax(death_team_totals) 
top_death_team = u_death_teams[top_death_team_idx]
top_death_runs = death_team_totals[top_death_team_idx]

# --- Output ---
print(f"Total runs scored in death overs (16-20): {int(death_runs_total)}")
print(f"Team with highest runs in death overs: {top_death_team}")
print(f"Runs scored by {top_death_team} in death overs: {int(top_death_runs)}")
```

```python
# Task 8: Highest Scoring Match

# 1. We already have 'runs_per_match' from Task 1, which is a list of (match_id, total_runs)
match_ids_list = [item[0] for item in runs_per_match]
total_runs_list = [item[1] for item in runs_per_match]

match_ids_arr = np.array(match_ids_list)
total_runs_arr = np.array(total_runs_list)

# 2. Find the index of the maximum value in the runs array
highest_runs_idx = np.argmax(total_runs_arr)

# 3. Retrieve the match_id and the run count using that index
highest_match_id = match_ids_arr[highest_runs_idx]
highest_total_runs = total_runs_arr[highest_runs_idx]

# 4. Output Format: (match_id, total_runs)
highest_scoring_match = (int(highest_match_id), int(highest_total_runs))

# --- Output ---
print("Highest Scoring Match (match_id, total_runs):")
print(highest_scoring_match) 
```

```python

# Task 9: Runs per Team per Match

# 1. Combine match_id and batting_team to create a unique ID for each innings
match_team_combined = np.char.add(np.char.add(match_ids.astype(str), "_"), batting_team)

# 2. Get unique innings identifiers and their mapping indices
unique_innings, inv_innings_indices = np.unique(match_team_combined, return_inverse=True)

# 3. Use bincount to sum batsman_runs for each unique innings
innings_runs = np.bincount(inv_innings_indices, weights=batsman_runs)

# 4. Create a list of tuples: (match_id, team_name, total_runs)
runs_per_team_per_match = []
for i in range(len(unique_innings)):
    m_id_str, team_name = unique_innings[i].split("_")
    runs_per_team_per_match.append((int(m_id_str), team_name, int(innings_runs[i])))

# --- Neat Output ---
print(f"{'Match ID':<10} | {'Team Name':<30} | {'Runs':<5}")
print("-" * 50)
for m_id, team, runs in runs_per_team_per_match[:10]:
    print(f"{m_id:<10} | {team:<30} | {runs:<5}")
```

```python
# Task 10: Match Winner Approximation

# 1. Get the list of all unique match IDs
unique_matches = np.unique(match_ids)

# 2. Extract match IDs from our 'unique_innings' identifiers for easy searching
innings_match_ids = np.array([int(s.split('_')[0]) for s in unique_innings])

# 3. List to store the results in the format (match_id, winner_team)
match_winners = []

for m in unique_matches:
    # Use boolean masking to find the indices of the two teams in this match
    match_indices = np.where(innings_match_ids == m)[0]
    
    if len(match_indices) > 0:
        # Get the runs scored by the teams in this match
        scores = innings_runs[match_indices]
        
        # Find the index of the highest score (np.argmax)
        winner_pos = np.argmax(scores)
        
        # Extract the team name of the winner from the unique_innings string
        winner_info = unique_innings[match_indices[winner_pos]]
        winner_team = winner_info.split('_')[1]
        
        # Append as a tuple (match_id, winner_team)
        match_winners.append((int(m), winner_team))

# --- Summary Table ---
print(f"\n{'Match ID':<10} | {'Estimated Winner':<30}")
print("-" * 45)
for m_id, winner in match_winners[:10]:
    print(f"{m_id:<10} | {winner:<30}")
```
